```python
seq = "AACGGCCGTTAAACTTG"

g_cont = seq.count("G")
c_cont = seq.count("C")

gc = (g_cont + c_cont) / len(seq)

print(gc)
```

```python
#we can use def to make it a function and use this in operations and apply it

def GC_content(seq):

    g_cont = seq.count("G")
    c_cont = seq.count("C")

    gc = (g_cont + c_cont) / len(seq)

    print(gc)
```

```python
GC_content("AACGGCCGTTAAACTTG") #this will print whatever we are doing without saving any variable.
```

```python
#insted of printing it will return

def GC_content(seq):

    g_cont = seq.count("G")
    c_cont = seq.count("C")

    gc = (g_cont + c_cont) / len(seq)

    return gc
```

```python
print("The GC content is", GC_content("AACGGCCGTTAAACTTG"))
```

```python
def squared(start=3, end=9):
    results = []
    for i in range(start, end):
        r = i**2
        results.append(r)
    return results
```

```python
squared(start=1, end=7)
```

```python
#you can look how the name of errors is and you can edit them let's say

x = 8
y = "a"

try:
    print(x/y)
except ZeroDivisionError:
    print("you can't divide by zero")
except TypeError:
    print("Wrong type dummy!")
else:
#even if the error exists the code keeps going so you have to add else to make it stop because otherwise it keeps going and mess your data.
    print("I kept going")
```
