## 5.2 Gobbler proteins 

Fix the function above, and use it to translate the fifteen mRNA sequences present in Turkey_transcripts_15_coding.fasta. Output a fasta file with the corresponding protein sequences. Give your new sequences the same sequence headers as in the transcript file, but instead of ending in "gbskey=CDS", end in "gbskey=AA". Note: The transcripts in our file are expressed as DNA sequences, so they have a Ts instead of Us for bases that would be an uracyl in teh actual RNA molecule.

```python
# Input and output files
input_file = "Turkey_transcripts_15_coding.fasta"
output_file = "Turkey_proteins_15.fasta"

# Open input and output files
#input file is just for reading, while output file is for writing
infile = open(input_file, "r")
outfile = open(output_file, "w")

#in python we can use readline function in order, in fasta file the first line is the header
#and the second line is the sequence.

# Read the first header
header = infile.readline().strip()
# Read the first DNA sequence
dna_sequence = infile.readline().strip()

# Read the rest of the file
for line in infile:
    
    line = line.strip()

    # If we find a new header
    if line.startswith(">"):

        # Convert DNA to mRNA
        mRNA = dna_sequence.replace("T", "U")

        # Translate mRNA using tour new function that we fixed in the previous exercise
        protein = get_amino_acids(mRNA)

        # Change CDS to AA in the header. It has to be "gbkey" and not "gbskey" otherwise it won't change anything
        new_header = header.replace("gbkey=CDS", "gbkey=AA")

        # Write header and protein to output file
        outfile.write(new_header + "\n")
        outfile.write(protein + "\n")

        # Save the new header
        header = line
         # Reset DNA sequence
        dna_sequence = ""

    else:

        # Add DNA to the current sequence
        dna_sequence = dna_sequence + line


# Translate the last sequence otherwise it will stop and not do it.
mRNA = dna_sequence.replace("T", "U")
protein = get_amino_acids(mRNA)

# Changing CDS to AA 
new_header = header.replace("gbkey=CDS", "gbkey=AA")

# Write the last header and protein
outfile.write(new_header + "\n")
outfile.write(protein + "\n")

# Closing files because in python we should do that everytime
infile.close()
outfile.close()


print(open(output_file).read())
```
