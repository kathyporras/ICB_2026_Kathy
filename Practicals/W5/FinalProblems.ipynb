# 4.2 Fizz buzz

The pseudocode for fizz buzz game would be:

x = numbers

if x divisible by 3, replace("fizz")
if x divisible by 5, replace("buzz")
if x divisible by 5 and 3, replace("fizzbuzz")

The actual code to represent this game is:

```python
x = 1

while x < 101:
    x = x + 1
    if x % 3 == 0:
        print(x, "fuzz")
    elif x % 5 == 0:
        print(x, "buzz")
```
