## 4.2 Fizz buzz

The pseudocode for fizz buzz game would be:

x = numbers from 1 to 100

if x divisible by 5 and 3: replace("fizzbuzz")

and if x divisible by 3: replace("fizz")

and if x divisible by 5: replace("buzz")


The actual code to represent this game is:

```python
#because counting in python start from 0
x = 0

#the first "if" is going to be the 3 and 5 together otherwise it will never come out, because with "elif" it 
#runs only if the previous one is false. Last print is to print all the rest of the numbers like would happen in the actual game.
while x < 100:
    x = x + 1
    if x % 5 == 0 and x % 3 == 0:
        print(x, "fizz buzz")
    elif x % 3 == 0:
        print(x, "fizz")
    elif x % 5 == 0:
        print(x, "buzz")
    else:
        print(x)
```

## 4.3 GC content from fasta

I had to google this one because I'm using some functions that we didn't use in class such as startswith() because of
the format of a .fasta document.

```python
with open ("/Users/kathy/Documents/VT/IntroCompBio/IntroBiolComp-2026/Python/DataFiles/Turkey_transcripts_15.fasta", "r") as infile, open("gc_content2.txt", "w") as outfile:
#ok here the challenge is that the fasta archives are data that is not tabulated so the example we had in class doesn't apply for this one
#so what I'm going to do it first is define my variables that I need and that are not tabulated. Since fasta starts with >gene and then an enter and keep
    #going. I need to remove that space so that the searching is possible.

    gene_id = ""
    seq = ""

    #removing spaces
    for line in infile:
        line = line.rstrip()

      #new command that I found that could work for fasta documents, which each line (new gene) starts with a >. So for each gene it will
        #calculate the GC content
        if line.startswith(">"):
            #if seq already contains a sequence, it means it finished reading the previous gene and it can calculate the GC content.
            if seq:
                gc = (seq.count("G") + seq.count("C")) / len(seq)
                outfile.write(gene_id + "\t" + str(gc) + "\n")

            #since there is multiple information per line in a fasta, I only want the first word which is the name of the gene
            gene_id = line[1:].split()[0]
            #start an empty sequence for a new gene
            seq = ""
        #this corresponds to the if startswith, so if the line doesnt start wih > it's still the sequence
        else:
            seq += line
    # # After the loop ends, this calculates the GC content of the last gene, because 
    #there is no new ">" after the last sequence to trigger the previous loops.
    if seq:
        gc = (seq.count("G") + seq.count("C")) / len(seq)
        outfile.write(gene_id + "\t" + str(gc) + "\n")                                                             
```
