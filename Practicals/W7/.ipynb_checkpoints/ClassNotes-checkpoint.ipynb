```python
#REGULAR EXPRESSIONS! they can work in any language 

#we are going to find a restriction enzyme
import re #regular expressions

dna = "ACCCAAATTGAAAAATGTGTGTGTAAACCCGGGAATTTTAACCACAACAGGGAGAATTCCCCAAACCCACGGTTAA"

#search function part of the package: two arguments pattern we wanna match and where.
match = re.search(r"GAATTC", dna)

match.group()
```

```python
#you can do it by iterations and find them all and you can add loops and everything

import re #regular expressions

dna = "ACCCAAATTGAAAAATGTGTGTGTAAACCCGGGAATTTTAACCACAACAGGGAGAATTCCCCAAACCCACGGTTAA"

#search function part of the package: two arguments pattern we wanna match and where.
#this are some functions you can use
match = re.search(r"TTC", dna)
match = re.findall(r"TTC", dna) #will name them all.

match.group()
```

```python
#all the enzymes I am interested in for example

#GTAGAC
#GTCGAC
#GTATAC
#GTCTAC

#So: GT[AC][GT]AC

import re #regular expressions

dna = "ACCCAAATTGAAAAATGTGTGTGTAAACCCGGGTATACAACCACAACAGGGAGAATTCCCCAAACCCACGGTTAA"

#search function part of the package: two arguments pattern we wanna match and where.
#this are some functions you can use
match = re.findall(r"GT[AC][GT]AC", dna) #here you are saying it could be any of them between the parenthesis.

print(match)
```

```python
#ANCHORS
mRNA = "TCTCTGGCCCAGGAGATTTAAAAAAA"

if re.search(r"A{4,}", mRNA):
    print("It has a poly-A")

else:
    print("No polyA tail!")
```

```python
#this code doesnt detect that is at the end so that's important

rand_seq = "TTTTAAATATAAAAAATTTGGCCAAT"

if re.search(r"A{4,}", rand_seq):
    print("It has a poly-A")

else:
    print("No polyA tail!")
```

```python
#so you can add this little symbol to say that is at the beginning or $ at the end.

rand_seq = "TTTTAAATATAAAAAATTTGGCCAAT"

if re.search(r"^A{4,}", rand_seq):
    print("It has a poly-A")

else:
    print("No polyA tail!")
```

```python
xMil = re.compile(r"GT[AC][GT]AC")

seq="TCCGTTAACAGGCTTCCAAGTCCAGTCGACAAAAATTT"

hits = re.finditer(xMil, seq)

for i in hits:
    print(i.start()+1, i.group()) #+1 because python starts counting with 0
```
