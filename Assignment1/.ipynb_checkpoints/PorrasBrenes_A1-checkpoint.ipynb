D. Write a script that returns a delimited table (e.g. comma-separated or tab-separated) with the number of times each individual was sampled, and saves it as Num_samples.csv. 

```python
#this is the script I biult for this 

#very similar to the previous one cut will focus just in the first column of the document and you have to indicate the doc in the function
#tail -n +2 is indicating to start in the second column because the first one are the headers
#sort -n will order the numbers of the ID
#uniq -c is the command in charge of counting the lines of each ID
# > tmp will save the counting in a temporal doc
cut -f1 "$1" | tail -n +2 | sort -n | uniq -c > tmp

#this means that is going to create a document with two columns separated by tabs
echo -e "maleID\tNum_samples" > Num_samples.csv

#and in here I added a loop that means that is going through the temporal doc that I created, collecting the info 
#and is going to append the results into the same document
while read count id
do
    echo -e "$id\t$count" >> Num_samples.csv
#whit will read all the lines of my temporal doc and then with rm tmp is going to remove that doc
done < tmp
rm tmp

#I ran it in the terminal:

bash CountID.sh Gesquiere2011_data.csv
head Num_samples.csv 

maleID	Num_samples
1	10
2	2
3	61
4	46
5	28
6	7
7	5
8	17
9	4
```

B. The current view on the origin of new species is that reproductive isolation plays a key role. How many times are the words "sterility" and "isolation" mentioned in the book? Hint: don't forget to make your search case-insensitive.

```python
#I ran the following in the terminal

#-i allows to find the word regardless it's lower or upper case
#number of "sterility"
grep -i "sterility" DarwinBook.txt | wc -l

#ANSWER: 95

#number for isolation
grep -i "isolation" DarwinBook.txt | wc -l

# ANSWER: 12
```

C. To get a better idea of the context in which Darwin mentioned these words, search for the 3 lines before and after each occurrence of the key words in point B, and save your results to a text file. Save one file for each key word.

```python
# I ran in the terminal -B and -A looks for the three lines before and after the word. Documents created with the comand > are in
#the folder

grep -i -B 3 -A 3 "sterility" DarwinBook.txt > sterility.txt
grep -i -B 3 -A 3 "isolation"  DarwinBook.txt > isolation.txt
```

## 3. Twelve days for Christmas

Write a python script that outputs the full lyrics of the song, as displayed in the file TwelveDaysOfChristmas.txt, available in the IntroBiolComp-2026/Assignment1/ directory.

```python
#understading this example
# Outer loop: A-C. 
for i in ["A", "B", "C"]:
    print("Iteration", i, "of the outer loop")
    
    #Inner loop: 0-4.
    for j in range(0,5):
        print("Iteration", j, "of the inner loop")
    print("\n")
```

```python
#making a nested loop

gifts = ["A Partridge in a Pear Tree", 
         "Two Turtle Doves, and",
         "Three French Hens",
         "Four Colly Birds",
         "Five Gold Rings",
         "Six Geese-a-Laying",
         "Seven Swans-a-Swimming",
         "Eight Maids-a-Milking",
         "Nine Ladies Dancing",
         "Ten Lords-a-Leaping",
         "Eleven Pipers Piping",
         "Twelve Drummers Drumming"]

days = ["First", 
        "Second", 
        "Third", 
        "Fourth",
        "Fifth",
        "Sixth", 
        "Seventh", 
        "Eighth",
        "Ninth", 
        "Tenth", 
        "Eleventh", 
        "Twelfth"]

for i in range(12):
    print("On the", days[i], "day of Christmas my true love gave to me:")

    #this one I had to google because it wasn't in order but backwards so it's -1 so that it includes the element 0 and -1 so that 
    #it goes reverse
    for j in range(i, -1, -1):
        print(gifts[j])

    print("\n")
```
